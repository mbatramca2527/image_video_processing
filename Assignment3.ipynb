```python
# WAP to draw a border around the input image ( or in other words pad the input image).
# Input the width of the border(in terms of pixel count) and the desired color of the border from the user.
# Try this for both binary and grayscale image
```

```python
import cv2 as cv
import numpy as np
```

```python
# Read binary image
img = cv.imread("image1.jpg", cv.IMREAD_GRAYSCALE)
```

```python
# Convert image to binary
_, img = cv.threshold(img, 127, 100, cv.THRESH_BINARY)
```

```python
# Input border width and color
width = int(input("Enter border width in pixels: "))
color = int(input("Enter border color (0 or 255): "))
```

```python
# Get image dimensions
rows, cols = img.shape
print("Number of rows:", rows)
print("Number of columns:", cols)
```

```python
# Create new array with space for border
bordered = np.full((rows + 2*width, cols + 2*width), color, dtype=np.uint8)
```

```python
# Copy original image into the center
bordered[width:width+rows, width:width+cols] = img
```

```python
# Display
display_img = cv.resize(bordered, (600, 400))
cv.imshow("Binary Image with Border", display_img)
```

```python
cv.waitKey(0)
cv.destroyAllWindows()
```

```python
#WAP for Image Compliment. Try this for both binary and grayscale image
```

```python
# Image complement
complement = 255 - img

# Display
cv.imshow("Original Binary Image", img)
cv.imshow("Binary Image Complement", complement)

# Save
cv.imwrite("binary_complement.png", complement)

cv.waitKey(0)
cv.destroyAllWindows()
```
