```python
import cv2 as cv
import matplotlib.pyplot as plt
```

```python
# read an image
image = cv.imread("image1.jpg")
```

```python
# display the image
plt.imshow(image)
```

```python
# write an image
cv.imwrite("output1.jpg",image)
```

```python
#converting image into different formats 
cv.imwrite("output1.png",image)
cv.imwrite("output1.tiff",image)
cv.imwrite("output1.bmp",image)
```

```python
#perform image resizing
resized = cv.resize(image,(400,400))
```

```python
plt.imshow(resized)
```

```python
# convert to grayscale image
gray = cv.cvtColor(resized,cv.COLOR_BGR2GRAY)
#display the image
plt.imshow(gray)
```

```python
# read and display a video
video = cv.VideoCapture("video1.mp4")

if not video.isOpened():
    print("Error: Cannot open video.")
else:
    while True:
        ret, frame = video.read()
        if not ret:
            break

        cv.imshow("video", frame)

        # Press 'q' to exit
        if cv.waitKey(25) & 0xFF == ord('q'):
            break

video.release()
cv.destroyAllWindows()
```

```python
# Extract frames from a video.
import os
video = cv.VideoCapture("video1.mp4")
output_folder = "frames"
os.makedirs(output_folder, exist_ok=True)

frame_count = 0

while True:
    ret, frame = video.read()
    if not ret:
        break

    frame_name = os.path.join(output_folder, f"frame_{frame_count}.jpg")
    cv.imwrite(frame_name, frame)
    frame_count += 1

video.release()
print(f"{frame_count} frames extracted and saved in '{output_folder}' folder.")
```
