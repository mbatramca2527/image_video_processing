```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read input image
img = cv2.imread("image2.jpg", 0)

# Log transformation
c = 1
log_img = c * np.log(1 + img)

# Convert to 8-bit range
log_img = np.uint8(255 * log_img / np.max(log_img))

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_img, cmap='gray')
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read input image
img = cv2.imread("image2.jpg", 0)

# Input gamma value
gamma = float(input("Enter the value of gamma: "))

# Normalize image
normalized = img / 255.0

# Power-law transformation
c = 1
gamma_img = c * np.power(normalized, gamma)

# Convert back to 8-bit
gamma_img = np.uint8(gamma_img * 255)

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_img, cmap='gray')
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read input image
img = cv2.imread("image2.jpg", 0)

# Get input range from user
r1 = int(input("Enter input range minimum: "))
r2 = int(input("Enter input range maximum: "))

# Get desired output range from user
s1 = int(input("Enter output range minimum: "))
s2 = int(input("Enter output range maximum: "))

# Create output image
output = np.zeros_like(img, dtype=np.uint8)

# Contrast stretching
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        pixel = img[i, j]

        if pixel < r1:
            output[i, j] = s1
        elif pixel > r2:
            output[i, j] = s2
        else:
            output[i, j] = ((pixel - r1) * (s2 - s1) /
                            (r2 - r1)) + s1

# Display images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(output, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis("off")

plt.show()
```
