```python
import cv2 as cv
import numpy as np
```

```python
#convert a grayscale image into a binary image.
# Read grayscale image
image = cv.imread("image1.jpg", cv.IMREAD_GRAYSCALE)

if image is None:
    print("Error: Image not found!")
    exit()

# -------------------------------
# Case 1: Threshold = Mean intensity
# -------------------------------
mean_val = np.mean(image)
ret, binary_mean = cv.threshold(image, mean_val, 255, cv.THRESH_BINARY)
print(ret)

cv.imshow("Binary (Mean Threshold)", binary_mean)
cv.imwrite("binary_mean.jpg", binary_mean)



# -------------------------------
# Case 2: User-defined threshold
# -------------------------------
threshold = int(input("Enter threshold value (0-255): "))

ret, binary_user = cv.threshold(image, threshold, 255, cv.THRESH_BINARY)
print(ret)

cv.imshow("Binary (User Threshold)", binary_user)
cv.imwrite("binary_user.jpg", binary_user)

cv.waitKey(0)
cv.destroyAllWindows()

```

```python
# Read color image
image = cv.imread("image1.jpg")

if image is None:
    print("Error: Image not found!")
    exit()

# Split channels (OpenCV uses BGR format)
B, G, R = cv.split(image)

# -------------------------------
# Case 1: Mean average method
# -------------------------------
gray_mean = ((R + G + B) / 3).astype(np.uint8)

cv.imshow("Grayscale (Mean Method)", gray_mean)
cv.imwrite("gray_mean.jpg", gray_mean)

# -------------------------------
# Case 2: User-defined weights
# -------------------------------
r_w = float(input("Enter weight for R: "))
g_w = float(input("Enter weight for G: "))
b_w = float(input("Enter weight for B: "))

# Validate weights
if (r_w + g_w + b_w) != 1:
    print("Error: Weights must sum to 1")
else:
    gray_weighted = (r_w * R + g_w * G + b_w * B).astype(np.uint8)

    cv.imshow("Grayscale (Weighted)", gray_weighted)
    cv.imwrite("gray_weighted.jpg", gray_weighted)

cv.waitKey(0)
cv.destroyAllWindows()
```
