```python
# Write a program to sharpen the input image using:  
# 1. Laplacian filter 
import cv2
import matplotlib.pyplot as plt

# Read input image
path = input("Enter image path: ")
img = cv2.imread(path)

if img is None:
    print("Image not found!")
else:
    # Apply Laplacian filter
    laplacian = cv2.Laplacian(img, cv2.CV_64F)

    # Convert Laplacian result to 8-bit
    laplacian = cv2.convertScaleAbs(laplacian)

    # Sharpen the image
    sharpened = cv2.addWeighted(img, 1.0, laplacian, -1.0, 0)

    # Display images
    plt.figure(figsize=(10, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
    plt.title("Laplacian Sharpened Image")
    plt.axis("off")

    plt.show()
```

```python
#2. Gradient operators of Sobel and Canny edge detector  
import cv2
import matplotlib.pyplot as plt

# Read input image
path = input("Enter image path: ")
img = cv2.imread(path)

if img is None:
    print("Image not found!")
else:
    # Convert image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Apply Sobel operators
    sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)

    # Calculate gradient magnitude
    sobel = cv2.magnitude(sobel_x, sobel_y)
    sobel = cv2.convertScaleAbs(sobel)

    # Apply Canny edge detector
    canny = cv2.Canny(gray, 100, 200)

    # Display results
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(gray, cmap='gray')
    plt.title("Original Grayscale")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(sobel, cmap='gray')
    plt.title("Sobel Edges")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(canny, cmap='gray')
    plt.title("Canny Edges")
    plt.axis("off")

    plt.show()
```

```python
#3.  Unsharp masking. To obtain blurred image, use the Gaussian filter of size 5x5 and sigma= 3 
import cv2
import matplotlib.pyplot as plt

# Read input image
path = input("Enter image path: ")
img = cv2.imread(path)

if img is None:
    print("Image not found!")
else:
    # Apply Gaussian blur
    blurred = cv2.GaussianBlur(img, (5, 5), 3)

    # Apply unsharp masking
    sharpened = cv2.addWeighted(img, 1.5, blurred, -0.5, 0)

    # Display images
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
    plt.title("Gaussian Blurred Image")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(cv2.cvtColor(sharpened, cv2.COLOR_BGR2RGB))
    plt.title("Unsharp Masked Image")
    plt.axis("off")

    plt.show()
```
